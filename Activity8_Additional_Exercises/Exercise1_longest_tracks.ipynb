```python
import pandas as pd
import matplotlib.pyplot as plt

# Gamitin ang "../" para hanapin ang CSV file sa parent folder
dataset_path = "../spotify_top_1000_tracks.csv"
df = pd.read_csv(dataset_path, encoding="utf-8")

print("Dataset loaded successfully!")
```

```python
longest = df.nlargest(10, 'duration_min')
print(longest[['track_name', 'artist', 'duration_min']])

plt.figure(figsize=(10, 6))
plt.barh(longest['track_name'], longest['duration_min'], color='indianred')
plt.xlabel('Duration (min)')
plt.ylabel('Track Name')
plt.title(f"Mary Gail Juarez (TUPM-26-1843)\nTop 10 Longest Songs")
plt.gca().invert_yaxis()
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
