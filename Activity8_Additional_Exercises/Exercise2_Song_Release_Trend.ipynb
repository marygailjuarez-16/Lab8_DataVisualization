```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. I-load ang dataset (Gamitin ang "../" kung ang notebook mo ay nasa loob ng isang subfolder)
dataset_path = "../spotify_top_1000_tracks.csv"
df = pd.read_csv(dataset_path, encoding="utf-8")

# 2. I-convert ang release_date at kunin ang taon
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
df['year'] = df['release_date'].dt.year

# 3. Subukan ang yearly counts
yearly_counts = df['year'].dropna().value_counts().sort_index()
print(yearly_counts.head())
```

```python
yearly_counts = df['year'].value_counts().sort_index()
cumulative_counts = yearly_counts.cumsum()

fig, ax1 = plt.subplots(figsize=(12, 6))
ax1.bar(yearly_counts.index, yearly_counts.values, color='royalblue', alpha=0.6, label='Yearly Count')
ax1.set_xlabel('Year')
ax1.set_ylabel('Number of Songs', color='royalblue')

ax2 = ax1.twinx()
ax2.plot(cumulative_counts.index, cumulative_counts.values, color='darkred', linewidth=2.5, label='Cumulative Growth')
ax2.set_ylabel('Cumulative Total', color='darkred')

plt.title(f"Mary Gail Juarez (TUPM-26-1843)\nSong Release Trend and Cumulative Growth")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
