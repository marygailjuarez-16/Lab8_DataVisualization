```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. I-load ang dataset (Gamitin ang "../" dahil nasa loob ka ng subfolder)
dataset_path = "../spotify_top_1000_tracks.csv"
df = pd.read_csv(dataset_path, encoding="utf-8")

# 2. I-convert ang release_date patungong year para mabasa ng animation
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
df['year'] = df['release_date'].dt.year

print("Dataset loaded successfully!")
```

```python
from matplotlib.animation import FuncAnimation, PillowWriter
import os

# 1. I-filter ang valid years at i-sort
animated_df = df.dropna(subset=['year']).sort_values('year')
years = sorted(animated_df['year'].unique())

# 2. I-setup ang figure para sa animation
fig, ax = plt.subplots(figsize=(10, 6))

def update(frame):
    ax.clear()
    current_year = years[frame]
    sub_df = animated_df[animated_df['year'] <= current_year]
    
    # Hexbin density plot para sa bawat taon
    ax.hexbin(sub_df['duration_min'], sub_df['popularity'], gridsize=20, cmap='coolwarm', mincnt=1, alpha=0.8)
    ax.set_xlim(animated_df['duration_min'].min(), animated_df['duration_min'].max())
    ax.set_ylim(0, 105)
    ax.set_xlabel('Duration (min)')
    ax.set_ylabel('Popularity Score')
    ax.set_title(f"Mary Gail Juarez (TUPM-26-1843)\nAnimated Hexbin Density Map (Year up to: {current_year})")

# 3. Buuin ang animation
ani = FuncAnimation(fig, update, frames=len(years), interval=200, repeat=False)

# 4. I-save bilang GIF file sa project folder
gif_path = os.path.abspath("duration_vs_popularity_animated.gif")
ani.save(gif_path, writer=PillowWriter(fps=5))
plt.close(fig)

print(f"Animated GIF successfully saved at: {gif_path}")
```
